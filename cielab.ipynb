```python
import numpy as np
import matplotlib.pyplot as plt

from skimage import io, color
```

```python
img = io.imread('~/Desktop/monet_sunrise.jpg')
# img = io.imread('~/Desktop/schendel_market.jpg')

io.imshow(img)
```

```python
img_lab = color.rgb2lab(img[:, :, :3])
plt.figure(figsize=(15, 5))
cmaps = ['gray', 'winter', 'RdBu']
for i in range(3):
    plt.subplot(1, 3, i+1)
    plt.imshow(img_lab[:, :, i], cmap=cmaps[i])
    plt.axis('off')
    plt.colorbar()
```
